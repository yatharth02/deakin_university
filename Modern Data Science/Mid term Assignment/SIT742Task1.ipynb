```python
def is_divisible(m, n):
  """Checks if m is divisible by n.

  Args:
    m: The dividend.
    n: The divisor.

  Returns:
    True if m is divisible by n, False otherwise.

  Exceptio:
    ValueError: when divisor is passed as 0.
  """

  if n == 0:
    raise ValueError("Cannot divide by zero.")

  return m % n == 0

# Test cases
print(f'1st test case where m => 20 and n => 5, result => {is_divisible(20, 5)}')
print(f'2nd test case where m => 20 and n => 3, result => {is_divisible(20, 3)}')
print(f'3rd test case where m => 49 and n => 7, result => {is_divisible(49, 7)}')

try:
  is_divisible(2, 0)
except Exception as e:
  print(f'4th test case for exception case where m => 2 and n => 0, result => {e}')
```

```python
def is_colliding(ball1, ball2):
  """Checks if two balls are colliding.

  Args:
    ball1: A tuple representing the first ball (x, y, r).
    ball2: A tuple representing the second ball (x, y, r).

  Returns:
    True if the balls are colliding, False otherwise.
  """

  x1, y1, r1 = ball1
  x2, y2, r2 = ball2

  distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

  return distance <= r1 + r2

# Test cases
print(f'2 balls with center at (0, 0) and (5, 0) and radius with 5 and 3 respectively, colliding result => {is_colliding((0, 0, 5), (5, 0, 3))}')
print(f'2 balls with center at (0, 0) and (8, 0) and radius with 5 and 3 respectively, colliding result => {is_colliding((0, 0, 5), (8, 0, 3))}')
print(f'2 balls with center at (0, 0) and (0, 0) and radius with 5 and 3 respectively, colliding result => {is_colliding((0, 0, 5), (0, 0, 3))}')
```

```python
def get_word_score(word):
    """
    Returns the score for a word. Assumes the word is a valid word.

    The score for a word is the product of two components:
    - The first component is the sum of the points for letters in the word.
    - The second component is the larger of: 1, or
      7*wordlen - 3*(20-wordlen), where wordlen is the
      length of the word. Letters are scored as in LETTER_VALUES.

    word: string
    returns: int >= 0
    """

    LETTER_VALUES = {
        'a': 1, 'b': 1, 'c': 5, 'd': 6, 'e': 1, 'f': 4, 'g': 2, 'h': 4, 'i': 1,
        'j': 8, 'k': 6, 'l': 1, 'm': 4, 'n': 1, 'o': 1, 'p': 3, 'q': 11, 'r': 1,
        's': 1, 't': 1, 'u': 1, 'v': 5, 'w': 4, 'x': 8, 'y': 6, 'z': 10, ' ': 100
    }

    word = word.lower()
    word_len = len(word)

    # First component: sum of the points for letters in the word
    letter_sum = sum(LETTER_VALUES.get(char, 0) for char in word)

    # Second component: 7*wordlen - 3*(20-wordlen)
    second_component = max(1, 7 * word_len - 3 * (20 - word_len))

    # Final score
    final_score = letter_sum * second_component

    return final_score

# Test cases
print(f'Score for "Australia" word => {get_word_score("Australia")}')
print(f'Score for "Deakin" word => {get_word_score("Deakin")}')
print(f'Score for "Yatharth Deoly" word => {get_word_score("Yatharth Deoly")}')
```

## Answer 2.1

```python
def graph(formula1, formula2, x_range):
    """Draws the intersections of the curve and the line.
       Also provides intersection points for the provided equations.

    Args:
      formula1: equation1 as lambda function type.
      formula2: equation2 as lambda function type.
      x_range: Array of numbers for x range.
    """

    # Converting x_range to array
    x_range = np.array(x_range)

    # Calculating different values of y for provided x values for different equations.
    y1 = formula1(x_range)
    y2 = formula2(x_range)

    # Plotting both equation in same plot
    plt.plot(x_range, y1, label='y = 10 - 2x')
    plt.plot(x_range, y2, label='y = 5 + 4x - x^2')

    # Define the variable
    x = sp.symbols('x')

    # Find the points of intersection by setting y1 equal to y2
    intersection_eq = sp.Eq(formula1(x), formula2(x))

    # Solve the equation for x
    intersection_points_x = sp.solve(intersection_eq, x)

    # Calculate the corresponding y-values using one of the original equations
    intersection_points = [(x_val.evalf(), formula1(x).subs(x, x_val).evalf()) for x_val in intersection_points_x]

    # Extract the x and y values
    x_intersections = [float(pt[0]) for pt in intersection_points]
    y_intersections = [float(pt[1]) for pt in intersection_points]

    # Plot intersection points
    plt.plot(x_intersections, y_intersections, 'ro', markersize=8)

    plt.title('The intersection of y = 10 − 2x and y = 5 + 4 x − x^2', y=-0.15)
    plt.legend(loc='upper right')
    plt.show()

    print()

    # Print intersection points
    for i, (x, y) in enumerate(intersection_points, 1):
        print(f"Intersection {i}: ({x:.2f}, {y:.2f})")

# Run the graph function
graph(lambda_exp1, lambda_exp2, val_range)
```
